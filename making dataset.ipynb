```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    'Age': [22, 25, 28, 30, 24, 35, 32, 40, 29, 38, 42, 45, 48, 50, 36],
    'Salary': [18000, 22000, 25000, 28000, 32000, 38000, 42000,
               45000, 48000, 55000, 65000, 72000, 85000, 95000, 110000]
}
```

```python

df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler=MinMaxScaler()
val = scaler.fit_transform(df)
val=np.round(val,2)
val
```

```python
dff = pd.DataFrame(val, columns=['Age', 'Salary'])
```

```python
from sklearn.cluster import KMeans
```

```python
km = KMeans(n_clusters=3, random_state=40)
km.fit_predict(dff)
```

```python
sse_data=[]
for i in range(1,11):
    km = KMeans(n_clusters=i, random_state=40)
    km.fit_predict(dff)
    sse_data.append(km.inertia_)
np.round(sse_data,4)
```

```python
plt.figure(figsize=(8,6))
plt.plot (range(1,11), sse_data, marker='o')
plt.xlabel("K")
```
