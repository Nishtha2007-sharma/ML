```python
import pickle

# 1. Define the data or object you want to save
data_to_save = {
    "name": "Alice",
    "age": 30,
    "skills": ["Python", "Machine Learning", "Data Science"]
}

# 2. Create and write to the .pkl file
with open("my_data.pkl", "wb") as file:
    pickle.dump(data_to_save, file)

print("Your .pkl file has been created successfully!")
```
